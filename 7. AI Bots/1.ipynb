````python
# 💀 Threatening Killer Voice Generator
import subprocess
import os
from pydub import AudioSegment, effects
import random
import time

def create_killer_voice(text, output_file="killer_voice.wav", intensity="extreme"):
    """
    Generate a threatening killer voice with menacing effects
    
    Intensity levels:
    - "menacing": Slightly threatening
    - "sinister": More intimidating
    - "killer": Very threatening
    - "extreme": Maximum fear factor
    """
    
    temp_file = "temp_killer_voice.wav"
    
    # Voice parameters for different intensity levels
    voice_configs = {
        "menacing": {"speed": 130, "pitch": 25, "amplitude": 180},
        "sinister": {"speed": 110, "pitch": 15, "amplitude": 200},
        "killer": {"speed": 90, "pitch": 10, "amplitude": 220},
        "extreme": {"speed": 70, "pitch": 5, "amplitude": 250}
    }
    
    config = voice_configs.get(intensity, voice_configs["extreme"])
    
    try:
        print(f"🎭 Generating {intensity} killer voice...")
        print(f"📝 Text: \"{text}\"")
        
        # Generate base voice with eSpeak (very slow and low pitch for menacing effect)
        subprocess.run([
            "espeak", 
            "-s", str(config["speed"]),    # Very slow speech
            "-p", str(config["pitch"]),    # Very low pitch (demonic)
            "-a", str(config["amplitude"]), # High amplitude
            "-v", "en+m3",                 # Male voice variant
            "-w", temp_file,
            text
        ], check=True)
        
        # Load and apply killer voice effects
        voice = AudioSegment.from_file(temp_file, format="wav")
        
        print("🔧 Applying killer voice effects...")
        
        # 1. Normalize first
        voice = effects.normalize(voice)
        
        # 2. Make it more menacing with frequency manipulation
        if intensity in ["killer", "extreme"]:
            # Deep, growling bass enhancement
            voice = voice.low_pass_filter(800)  # Remove high frequencies
            voice = voice.high_pass_filter(100) # Keep deep bass
            
            # Add slight distortion for more menacing effect
            voice = voice + 12  # Boost volume for distortion
            voice = effects.compress_dynamic_range(voice, threshold=-20.0, ratio=4.0)
            
        elif intensity == "sinister":
            voice = voice.low_pass_filter(1000)
            voice = voice.high_pass_filter(150)
            voice = voice + 8
            
        else:  # menacing
            voice = voice.low_pass_filter(1200)
            voice = voice.high_pass_filter(200)
            voice = voice + 5
        
        # 3. Add echo/reverb effect for horror atmosphere
        if intensity in ["killer", "extreme"]:
            # Create echo by overlaying delayed versions
            echo1 = voice.apply_gain(-8)  # Quieter echo
            echo2 = voice.apply_gain(-12) # Even quieter echo
            
            # Overlay echoes at different delays
            voice = voice.overlay(echo1, position=150)  # 150ms delay
            voice = voice.overlay(echo2, position=300)  # 300ms delay
        
        # 4. Add slight tremolo/vibrato for unnerving effect
        if intensity == "extreme":
            # Create a subtle tremolo by varying volume
            tremolo_voice = AudioSegment.empty()
            chunk_size = 50  # 50ms chunks
            
            for i in range(0, len(voice), chunk_size):
                chunk = voice[i:i+chunk_size]
                # Vary gain slightly for tremolo effect
                gain_variation = random.uniform(-2, 2)
                chunk = chunk.apply_gain(gain_variation)
                tremolo_voice += chunk
            
            voice = tremolo_voice
        
        # 5. Final processing
        voice = effects.normalize(voice)
        voice = effects.compress_dynamic_range(voice, threshold=-15.0, ratio=3.0)
        
        # 6. Add silence padding for dramatic effect
        silence_before = AudioSegment.silent(duration=500)  # 0.5 sec silence
        silence_after = AudioSegment.silent(duration=1000)  # 1 sec silence
        voice = silence_before + voice + silence_after
        
        # Export the final killer voice
        voice.export(output_file, format="wav")
        
        # Cleanup
        if os.path.exists(temp_file):
            os.remove(temp_file)
        
        # File info
        file_size = os.path.getsize(output_file)
        duration = len(voice) / 1000.0
        
        print(f"💀 Successfully created killer voice: {output_file}")
        print(f"📊 File size: {file_size:,} bytes")
        print(f"⏱️  Duration: {duration:.2f} seconds")
        print(f"🎭 Intensity level: {intensity.upper()}")
        
        return True
        
    except Exception as e:
        print(f"❌ Error creating killer voice: {e}")
        return False

# College Rivals 3 threatening message
college_rivals_message = (
    "You can't hide. You can't run. All that's left… is to fight. "
    "College Rivals 3. 28th August. E3 Palm Court. "
    "Come prove yourself… or fall."
)

print("💀 COLLEGE RIVALS 3 - KILLER VOICE GENERATOR")
print("=" * 60)
print(f"📅 Event Date: 28th August")
print(f"📍 Location: E3 Palm Court")
print(f"🎯 Message: {college_rivals_message}")
print("=" * 60)

# Generate different intensity levels of the killer voice
intensities = ["menacing", "sinister", "killer", "extreme"]

for intensity in intensities:
    filename = f"college_rivals_3_{intensity}_voice.wav"
    print(f"\n🎭 Creating {intensity.upper()} version...")
    
    if create_killer_voice(college_rivals_message, filename, intensity):
        print(f"✅ {filename} created successfully!")
    else:
        print(f"❌ Failed to create {filename}")
    
    # Small delay between generations
    time.sleep(0.5)

print(f"\n🎉 All killer voice variations created!")
print("🎧 Use the cells below to play the generated voices.")
````

````python
# 👤 PURE HUMAN THREATENING VOICE (NO ROBOTIC EFFECTS)
import subprocess
import os
from pydub import AudioSegment, effects

def create_human_threatening_voice(text, output_file="human_killer_voice.wav", voice_style="killer"):
    """
    Generate a threatening human voice WITHOUT any robotic effects
    Focus on natural human voice with menacing tone
    
    Voice styles:
    - "intimidating": Slightly threatening human voice
    - "menacing": More threatening with natural effects
    - "killer": Maximum human threat level
    """
    
    temp_file = "temp_human_voice.wav"
    
    # Voice parameters for natural human threatening voice
    voice_configs = {
        "intimidating": {"speed": 140, "pitch": 30, "amplitude": 190},
        "menacing": {"speed": 120, "pitch": 20, "amplitude": 210},
        "killer": {"speed": 100, "pitch": 15, "amplitude": 230}
    }
    
    config = voice_configs.get(voice_style, voice_configs["killer"])
    
    try:
        print(f"👤 Generating PURE HUMAN {voice_style} voice...")
        print(f"📝 Text: \"{text}\"")
        print("🚫 NO robotic effects - Pure human threatening voice")
        
        # Generate base voice with eSpeak (natural human settings)
        subprocess.run([
            "espeak", 
            "-s", str(config["speed"]),    # Controlled speech speed
            "-p", str(config["pitch"]),    # Lower pitch for threat
            "-a", str(config["amplitude"]), # Clear amplitude
            "-v", "en+m3",                 # Male voice for threat
            "-w", temp_file,
            text
        ], check=True)
        
        # Load audio
        voice = AudioSegment.from_file(temp_file, format="wav")
        
        print("🎚️  Applying NATURAL human voice enhancements...")
        
        # ONLY natural human voice enhancements (NO robotic effects)
        
        # 1. Normalize volume for clarity
        voice = effects.normalize(voice)
        
        # 2. Natural frequency response (preserve human voice characteristics)
        if voice_style == "killer":
            # Enhance natural bass for more threatening tone
            # Keep full human frequency range but emphasize lower frequencies
            voice = voice.low_pass_filter(3000)   # Keep human voice range
            voice = voice.high_pass_filter(80)    # Remove very low rumble
            
        elif voice_style == "menacing":
            # Slight natural enhancement
            voice = voice.low_pass_filter(3500)
            voice = voice.high_pass_filter(90)
            
        else:  # intimidating
            # Minimal processing to keep natural sound
            voice = voice.low_pass_filter(4000)
            voice = voice.high_pass_filter(100)
        
        # 3. Natural dynamic range compression (like human speech)
        voice = effects.compress_dynamic_range(voice, threshold=-12.0, ratio=2.0)
        
        # 4. Slight natural reverb (like speaking in a room)
        if voice_style in ["menacing", "killer"]:
            # Add very subtle natural echo (like speaking in a large room)
            natural_echo = voice.apply_gain(-15)  # Very quiet echo
            voice = voice.overlay(natural_echo, position=80)  # Short natural delay
        
        # 5. Final natural processing
        voice = effects.normalize(voice)
        
        # 6. Add natural pauses for dramatic effect
        short_pause = AudioSegment.silent(duration=300)   # 0.3 sec pause
        long_pause = AudioSegment.silent(duration=800)    # 0.8 sec pause
        
        voice = short_pause + voice + long_pause
        
        # Export the pure human threatening voice
        voice.export(output_file, format="wav")
        
        # Cleanup
        if os.path.exists(temp_file):
            os.remove(temp_file)
        
        # File info
        file_size = os.path.getsize(output_file)
        duration = len(voice) / 1000.0
        
        print(f"👤 Successfully created HUMAN threatening voice: {output_file}")
        print(f"📊 File size: {file_size:,} bytes")
        print(f"⏱️  Duration: {duration:.2f} seconds")
        print(f"🎭 Style: {voice_style.upper()} (100% human, 0% robotic)")
        
        return True
        
    except Exception as e:
        print(f"❌ Error creating human threatening voice: {e}")
        return False

# College Rivals 3 message for PURE HUMAN threatening voice
message = (
    "You can't hide. You can't run. All that's left… is to fight. "
    "College Rivals 3. 28th August. E3 Palm Court. "
    "Come prove yourself… or fall."
)

print("👤 COLLEGE RIVALS 3 - PURE HUMAN THREATENING VOICE")
print("🚫 NO ROBOTIC EFFECTS - 100% NATURAL HUMAN VOICE")
print("=" * 65)
print(f"📅 Event: College Rivals 3")
print(f"📅 Date: 28th August") 
print(f"📍 Venue: E3 Palm Court")
print("=" * 65)

# Generate different styles of human threatening voice
human_styles = ["intimidating", "menacing", "killer"]

for style in human_styles:
    filename = f"college_rivals_3_human_{style}.wav"
    print(f"\n👤 Creating HUMAN {style.upper()} voice...")
    
    if create_human_threatening_voice(message, filename, style):
        print(f"✅ {filename} created successfully!")
        
        # Play the generated voice immediately
        import IPython.display as ipd
        print("🎧 Playing generated voice:")
        display(ipd.Audio(filename, autoplay=False))
        print("-" * 50)
    else:
        print(f"❌ Failed to create {filename}")

print(f"\n🎉 All PURE HUMAN threatening voices created!")
print("👤 These voices sound like a real threatening human - NO robotic effects!")
print("🎭 Perfect for creating genuine fear without artificial sound!")
````
