## Intitial Setup

```python
# Load Environment variables
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")
```

## Testing using Groq

### Approach 1

```python
# Load LLM model for chatting 
from langchain.chat_models import init_chat_model
model=init_chat_model("openai/gpt-oss-20b", model_provider="groq")
model
```

```python
response = model.invoke("Write me an essay on AI")
response
```

```python
print(response.content)
```

### Approach 2

```python
from langchain_groq import ChatGroq
model = ChatGroq(model="openai/gpt-oss-20b")
model
```

```python
response = model.invoke("Write me an essay on AI")
print(response.content)
```

## Testing using Gemini

### Approach 1

```python
os.environ["GEMINI_API_KEY"]=os.getenv("GEMINI_API_KEY")
from langchain.chat_models import init_chat_model
model=init_chat_model("google_genai:gemini-3-flash-preview")
model
```

```python
response = model.invoke("Write me an essay on AI")
print(response.content)
```

### Approach 2

```python
from langchain_google_genai import ChatGoogleGenerativeAI
model = ChatGoogleGenerativeAI(model="gemini-3-flash-preview")
model
```

```python
response = model.invoke("Write me an essay on AI")
print(response.content)
```

## Streaming

Displaying while generating response

```python
model.stream("Write me a 1000 words essay on AI")
```

```python
for chunk in model.stream("Write me a 1000 words essay on AI"):
    print(chunk.text, end="", flush=True)
```
