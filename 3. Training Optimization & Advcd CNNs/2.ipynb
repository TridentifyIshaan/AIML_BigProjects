# Advanced CNNs

Convolutional Neural Network- CNN is a distinct kind of Neural Networks, which is proved to be the best technique to handle several interesting problems related to Computer Vision and Image Processing.

The learning capability of the deep CNN is mainly because of employing the multiple feature extraction techniques that can automatically extract the patterns from the data.   This ability of CNN makes it more powerful and effective CNN.

The voluminous data that is being gathered and the developments in the hardware technology has influenced the new advanced architecture for CNN. Top industries such as Google, AT&T, Amazon, Tesla, Facebook and Microsoft are exploring new architectures of CNN through dynamic research group.

## History of deep CNN architectures


Whose origin rooted in Fukushima’s Neocognitron
1. ConvNet:

   - First multilayered CNN
   - A supervised training
   - A backpropagation algorithm
   - An effective outcome was observed when implemented for handwritten digit recognition and recognition of zip code problems

2. LeNet:

   - Improved form of ConvNet
   - Used in a document processing for character recognition problems
   - Proved to perform optimal  in optical character recognition (OCR)and Biometric-fingerprint recognition
   - Commercial use in Bank application and ATMs during 1993 and 1996
   - Limitation: didn’t accomplish better results for other type image recognition problem

3. AlexNet:

   - The main innovation in 2012
   - The Idea of basic hyperparameter like stride, filter dimensions, padding, and other was presented for each layer to leverage the image recognition problem

4. GoogleNet:

   - It was designed based on the concept of minor or simple networks
   - To have an effective computation, the conventional convolution operation was replaced with point-wise group convolution operation.
   - The GoogleNet architecture mainly focused to attain high accuracy with a minimum computational effort
   - Limitations: Since the topology is Heterogeneous in nature, each module requires customization. Feature representation causes the drastic reduction in feature space size in the subsequent layer resulting in  information loss

5. VGG and ResNet:

- They are proved to be successful in addressing the challenging problems like image recognition and localization. For example: object segmentation based on instance and semantic, analysis of the scene, localization of the scene, etc.

## Applications of CNN

Some of the fascinating areas where CNN are used is as discussed below:
1. Computer vision and other associated applications using CNN

    - Face recognition: Even though face recognition task was one of the well-known problems addressed earlier, the photos taken under variations in lighting, variation in position, and diversified facial expressions makes this task difficult
    - CNN is effectively used in Action/ activity recognition

2. Natural language processing based on CNN

    Due to the recurrent nature of the RNNs, and due to availability of memorizing the sequence option in LSTM, they are found to be very appropriate for NLP applications. However, language modeling, and analysis of the text can be effectively handled by CNN also.

3. Object detection and segmentation

    Identification of various objects in the input images is called Object detection. A variant of CNN called R-CNN is found to be widely used for this task

4. Image classification

    Medical images are always posing challenges to machine learning algorithms. Particularly few associated tasks like the cancer diagnosis using histopathological images. CNN techniques become the best choice to handle medical images

5. Speech recognition

    Speech processing is very much essential in recognizing the vocabulary, recognition of emotion associated with speech which communicating over the Phone. CNN become quite useful architecture to handle speech recognition

6. Video processing

    Since CNN architecture are flexible and suitable to both temporal/time series data and spatial information, so, it is widely used in video processing.

## Challenges in CNN:

- Deep CNN architectures s is generally a black box model. So, they are not interpretable and explainable. Also, verification of these models is difficult task.
- Training the CNN with noisy training set or noisy images results in classification error.
- Hyper-parameter tuning in CNN will highly influences the performance of the model. An improper hyperparameter selection can affect the overall performance of a CNN.

### Importing Libraries

```python
import numpy as np
from keras.datasets import mnist
import matplotlib.pyplot as plt
from keras.models import Sequential
from keras.utils import to_categorical
from keras.layers import Dense, Dropout, Activation, Flatten
from keras.layers import Conv2D, MaxPooling2D
from keras.layers import BatchNormalization
```

### Loading Dataset

```python
(x_train, y_train), (x_test, y_test) = mnist.load_data()
print(x_train.shape)
print(x_test.shape)
```

### Reshaping, Normalization, etc..

```python
element = 200
plt.imshow(x_train[element])
plt.show()
print("Label for the element", element,":", y_train[element])
x_train = x_train.reshape((-1, 28*28))
x_test = x_test.reshape((-1, 784))
print(x_train.shape)
print(x_test.shape)
x_train = x_train / 255
x_test = x_test / 255
```

### Creating Model

```python
model = Sequential()
# 1st Convolutional Layer 
model.add(Conv2D(filters = 96, input_shape = (224,224, 1),kernel_size = (11, 11), strides = (4, 4), padding = 'valid')) 
model.add(Activation('relu')) 
# Max-Pooling  
model.add(MaxPooling2D(pool_size = (2, 2),strides = (2, 2), padding = 'valid')) 
# Batch Normalisation 
model.add(BatchNormalization())

# 2nd Convolutional Layer 
model.add(Conv2D(filters = 256, kernel_size = (3, 3),strides = (1, 1), padding = 'valid')) 
model.add(Activation('relu')) 
# Max-Pooling 
model.add(MaxPooling2D(pool_size = (2, 2), strides = (2, 2),padding = 'valid')) 
# Batch Normalisation 
model.add(BatchNormalization()) 
  
# 3rd Convolutional Layer 
model.add(Conv2D(filters = 384, kernel_size = (3, 3),strides = (1, 1), padding = 'valid')) 
model.add(Activation('relu')) 
# Batch Normalisation 
model.add(BatchNormalization()) 

# 4th Convolutional Layer 
model.add(Conv2D(filters = 384, kernel_size = (3, 3),strides = (1, 1), padding = 'valid')) 
model.add(Activation('relu')) 
# Batch Normalisation 
model.add(BatchNormalization()) 

# 5th Convolutional Layer 
model.add(Conv2D(filters = 256, kernel_size = (3, 3),strides = (1, 1), padding = 'valid')) 
model.add(Activation('relu')) 
# Max-Pooling 
model.add(MaxPooling2D(pool_size = (2, 2), strides = (2, 2), padding = 'valid')) 
# Batch Normalisation 
model.add(BatchNormalization()) 

# Flattening 
model.add(Flatten()) 
```

### Dense & Output Layers

```python
# 1st Dense Layer 
model.add(Dense(4096)) 
model.add(Activation('relu')) 
# Add Dropout to prevent overfitting 
model.add(Dropout(0.4)) 
# Batch Normalisation 
model.add(BatchNormalization()) 

# 2nd Dense Layer 
model.add(Dense(4096)) 
model.add(Activation('relu')) 
# Add Dropout 
model.add(Dropout(0.4)) 
# Batch Normalisation 
model.add(BatchNormalization()) 

# Output Softmax Layer 
model.add(Dense(10)) 
model.add(Activation('softmax'))
```

### Compile, Fit, Evaluate, Predict

```python
import tensorflow as tf

# Helper function to resize and preprocess images
def preprocess(image, label):
    image = tf.image.resize(image, [224, 224])
    if image.shape.rank == 2:  # If missing channel dimension
        image = tf.expand_dims(image, -1)
    return image, tf.one_hot(label, 10)

# Reshape x_train and x_test to (num_samples, 28, 28, 1)
x_train_reshaped = x_train.reshape(-1, 28, 28, 1)
x_test_reshaped = x_test.reshape(-1, 28, 28, 1)

# Create tf.data.Dataset objects for training and testing
train_ds = tf.data.Dataset.from_tensor_slices((x_train_reshaped, y_train))
train_ds = train_ds.map(preprocess, num_parallel_calls=tf.data.AUTOTUNE)
train_ds = train_ds.batch(64).prefetch(tf.data.AUTOTUNE)

test_ds = tf.data.Dataset.from_tensor_slices((x_test_reshaped, y_test))
test_ds = test_ds.map(preprocess, num_parallel_calls=tf.data.AUTOTUNE)
test_ds = test_ds.batch(64).prefetch(tf.data.AUTOTUNE)

# Compile the model
model.compile(optimizer='Adam', loss='categorical_crossentropy', metrics=['accuracy'])

# Fit the model
model.fit(train_ds, epochs=10, shuffle=True)

# Evaluate the model
eval = model.evaluate(test_ds)
print('eval:', eval)

# Predictions (get a batch from test_ds)
for images, labels in test_ds.take(1):
	predictions = model.predict(images)
	print('Predicted class:', np.argmax(predictions[0]))
	plt.imshow(tf.squeeze(images[0]).numpy(), cmap='gray')
	plt.show()
```
